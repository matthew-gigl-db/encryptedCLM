```sql
WITH gndr_diag AS (
    SELECT
        e.patient_gender,
        d.diagnosis_code,
        COUNT(e.patient_id) AS patient_count
    FROM
        mgiglia.hv_claims.enrollment e
    JOIN mgiglia.hv_claims.diagnosis d ON e.patient_id = d.patient_id
    WHERE
      array_contains(split(:gender, ","), e.patient_gender)
    GROUP BY
        e.patient_gender,
        d.diagnosis_code
)
SELECT 
    *
FROM 
    gndr_diag
ORDER BY
    patient_count DESC;
```
